## Advanced Slicing with Step

切片（slicing）的一般格式为：

```python
[start:end:step]
```

其中：

- `start`：切片开始的索引（包含该位置，inclusive）
- `end`：切片结束的索引（不包含该位置，exclusive）
- `step`：索引每次移动的步长（默认值为 `1`）

### 更准确地理解 step

❌ 不推荐理解为：

> 跳过多少个字符（how many characters to skip）

因为这种说法容易产生误解。

✅ 更推荐理解为：

> 索引每次移动多少个位置（how many index positions to move each time）

### 示例

```python
s = "abcdefg"
```

索引：

```text
a b c d e f g
0 1 2 3 4 5 6
```

#### step = 1

```python
s[0:7:1]
```

访问索引：

```text
0 → 1 → 2 → 3 → 4 → 5 → 6
```

结果：

```python
'abcdefg'
```

---

#### step = 2

```python
s[0:7:2]
```

访问索引：

```text
0 → 2 → 4 → 6
```

结果：

```python
'aceg'
```

---

#### step = 3

```python
s[0:7:3]
```

访问索引：

```text
0 → 3 → 6
```

结果：

```python
'adg'
```

---

### 负数 step

#### step = -1

```python
s[::-1]
```

访问索引：

```text
6 → 5 → 4 → 3 → 2 → 1 → 0
```

结果：

```python
'gfedcba'
```

即字符串反转。

---

### 记忆口诀

> start 是起点  
> end 是终点（不包含）  
> step 是索引每次移动的步长

或者更简单地记：

> **step 表示“每次移动几步”，不是“跳过几个字符”。**
