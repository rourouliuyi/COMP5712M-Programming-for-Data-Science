## (2) Basic data types

```python
# number
# integer, float
x = 5
y = 6
print(x + y)
```

```python
# string
module = 'Programming for Data Science'
print(module)
print(module[0:11])   # slicing
print(module.upper()) # string method upper()
print(module.split()) # splitting a string into parts
print('Data' in module) # checking for substring
```

```python
# boolean

print(5 > 9)
```
