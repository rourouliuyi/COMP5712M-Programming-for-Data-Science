```python
# conditional if-else

account_balance = 5000
spending = 8000

if (account_balance >= spending):
    account_balance -= spending
else:
    print("Insufficient fund!")
```

```python
# Conditional if-elif-if

spend = 1000

if spend >= 1500: 
    print('Platinum')
elif spend >= 500: 
    print('Gold')
elif spend > 0: 
    print('Silver')
else: 
    print('Inactive')
```

## Loops

```python
# for loop
votes = ['candidate1', 'candidate1', 'candidate2', 'candidate3', 'candidate4', 'candidate5',
         'candidate1', 'candidate3', 'candidate4', 'candidate2', 'candidate1', 'candidate3',
         'candidate4', 'candidate2', 'candidate3', 'candidate4']
count_votes = {}

for vote in votes:
    if vote in count_votes:
        count_votes[vote] += 1
    else:
        count_votes[vote] = 1
print(count_votes)
```

```python
# while loop

countdown = 5

while countdown > 0:
    print(countdown)
    countdown -= 1
    
print("Liftoff!")
```

```python
# nested loops

for i in range(1, 6): # Outer loop for rows 
    for j in range(1, 6): # Inner loop for columns   
        print(f"{i*j:2}", end=" ") # Print product, formatted   
    print() # New line after each row
```

## Comprehensions

```python
# List comprehension
animals = ['dog', 'cat', 'hamster', 'goldfish']
not_fish = [item for item in animals if item != 'goldfish']

not_fish
```

```python
# set comprehension
raw_subjects = ["Math", "science", "MATH", "History", "science"]
clean_subjects = {s.lower() for s in raw_subjects}

clean_subjects
```

```python
# Dictionary comprehension
user_status = {"Alice": "active", "Bob": "inactive", "Charlie": "active", "David": "expired"}
active_users = {user: status for user, status in user_status.items() if status == "active"}

active_users
```

## Defining and calling user defined functions

```python
# defining a function

def clean_text_data(text: str) -> str:
    """Converts a string to lowercase and removes leading and trailing whitespace."""
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = text.strip()
    
    return text
```

```python
# calling a function

clean_text = clean_text_data(" Jason Borne ")

clean_text
```

```python
# parameters and keyword arguments

def is_christmas(month, day):
    """Checks if a given month and day correspond to Christmas Day (December 25)."""
    return month == 12 and day == 25
```

```python
# values are passed in the same order, month=12, day=24
print(is_christmas(12, 24))
```

```python
# values are passed with keyword

print(is_christmas(day=25, month=12))
```
