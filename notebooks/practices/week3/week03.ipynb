## 3.1 Object-oriented programming (OOP) in Python

### Python is an object-oriented programming language.

### In OOP, we model our programs around objects that represent real-world things or concepts. 

### Each object has data (called attributes) and behaviours (called methods) and we have seen examples of strings and string methods.

### Sometimes in your project, you need to create specific objects and you can do that in Python, as in many other programming languages.

- ### to create an object, called class in Python, use the `class` keyword

```python
# define a class called TextAnalyser

class TextAnalyser:
    """ A simple class to analyze text data and compute basic NLP metrics. """

    def __init__(self, raw_text: str):
        self.raw_text = raw_text
        self.words = []
        self.tokenise()
        self.word_length = self.get_word_count()

    def tokenise(self):
        """A helper method to clean punctuation and split text into a list of words."""
        
        cleaned_text = self.raw_text.strip().lower()
        for punctuation in [".", ",", "!", "?", ";", ":"]:
            cleaned_text = cleaned_text.replace(punctuation, "")

        self.words = cleaned_text.split()

    def get_word_count(self) -> int:
        """Return the total number of words in the text."""
        return len(self.words)

    def get_word_frequencies(self) -> dict:
        """Calculate how many times each word appears using a dictionary."""
        frequencies = {}
        for word in self.words:
            if word in frequencies:
                frequencies[word] += 1
            else:
                frequencies[word] = 1
        return frequencies
```

### A few things here:

- ### `__init__( )` method - dunder init automatically called when creating an instance of the class
- ### `self` - refer to specific instance of the class, allowing access to its attributes and methods

### 1. To create an instance of the class, write the name of the class followed by parenthesis ( ) and parameters for `__init__( )`, without need to pass anything to `self`, that's automatic.

### `<ClassName>(<parameters for __init__>)`

### 2. To use the methods belonging to objects, use the . operator. This is the case with `"hello world".upper()`, with `upper()` being a string method, and `"hello world"` is an instance of string.

```python
# let's create an instance of TextAnalyser

sample_data = "Data science is amazing. Python is amazing for data science!    "

# instantiate the class
analyser = TextAnalyser(sample_data)
```

```python
# access instance's attributes
word_count = analyser.word_length
word_count
```

```python
# access instance's methods

word_frequency = analyser.get_word_frequencies()
print(f"Words frequency: {word_frequency}")
```

## 3.2 File Input/Output

### When working on data science projects, you will work with different types of data in a variety of formats. The common ones are TXT, CSV, and JSON files. Many of you will work with CSV files for the group project.

### Python offers many options to open file to read and write, from the conventional to specific modules.

### To open a file to read, we can use the `open()` function with the `with` context manager.

### Example 1: Sales Data Analysis

Given data in `sales_data.csv`, we want to analyse the sales data for:

- Total sales amount

```python
with open("sales_data.csv", "r") as f:
    data = f.readlines()

    # as you might suspect, there's \n at the end of each line
    # you can remove those with a string method
    data = [line.strip() for line in data]

    # we can do more such as split those into individual item
    data = [line.split(',') for line in data]

    for line in data:
        print(line)

    # you might want to convert Sales_Amount and Units_Sold to integer so that
    # you can sum them up

    converted_data = [[line[0], line[1], int(line[2]), int(line[3]), line[4]] for line in data[1:]]

    print("After conversion")
    for line in converted_data:
        print(line)

    # calculate sum Sales_amount
    total_sales = sum(line[2] for line in converted_data)
    print(total_sales)
```

### As you can see, we are doing a lot of things after opening the file. We have learned about function last week, it's a good practice to write modular code.

```python
# define a separate function for data cleaning

def clean_sales_data(data: list) -> list: 
    """
        docstring here
        Input:
        Output:
        Usage:
    """
    data = [line.strip().split(',') for line in data]
    converted_data = [[line[0], line[1], int(line[2]), int(line[3]), line[4]] for line in data[1:]]
    return converted_data
```

```python
# define a separate function to calculate total sales

def calculate_total_sales(data: list) -> int:
    """
        docstring here
        Input:
        Output:
        Usage:
    """
    return sum(line[2] for line in data)
```

```python
# define a separate function to open file and call other functions for a complete program

with open("sales_data.csv", "r") as f:
    data = f.readlines()

    data = clean_sales_data(data)
    print(calculate_total_sales(data))
```

## Example 2: Analyse student data to check if they are eligible to graduate

```python
# define function to clean student data

def clean_student_data(data:list) -> list:
    """
    Cleans and converts student data from comma-separated strings into lists.

    Strips whitespace from each input line, splits the line by commas,
    preserves the first field as a string, converts non-empty remaining
    fields to integers, and ignores empty fields.

    Args:
        data (list): A list of strings containing comma-separated student data.

    Returns:
        list: A list of lists containing student names (strings) followed
              by their numerical data (integers).
    """
    
    data = [line.strip().split(',') for line in data]

    new_list = []
    for line in data:
        temp_list = []
        temp_list.append(line[0])
        for i in line[1:]:
            if i != "":
                temp_list.append(int(i))
            # what if ""?
            else:
                temp_list.append(None)
        new_list.append(temp_list)

    # list comprehension
    # new_list = [[line[0]] + [int(i) if i != "" else None for i in line[1:]] for line in data]

    return new_list
    
```

```python
# return a list of obtained credits

def obtained_credits(data: list) -> list:
    """
    Check the marks for each student and return a list of obtained credits
    
    Assumptions:
        - the first item is student id
        - each module is 15 credits, except the last module which is 60 credits
        - passing mark for each module: 50
    """
    
    obtained_credit = [15 if module and module>=50 else 0 for module in data[1:-1]]

    if data[-1] and data[-1] >= 50:
        obtained_credit.append(60)
    else:
        obtained_credit.append(0)

    # list comprehension
    # obtained_credit = [15 if module and module >= 50 else 0 for module in data[1:-1]] + [60 if data[-1] and data[-1] >= 50 else 0]

    # testing
    # print(data)
    # print(f"obtained credits: {obtained_credit}, total credits: {sum(obtained_credit)}")
    
    return obtained_credit
```

```python
# Check if sufficient credits is achieved for graduation

def credits_achieved(obtained_credit:list) -> bool:
    """
    Check list of obtained credits for individual student and determine weather sufficient credits obtained.
    Assumptions:
        - required credit: 150/180

    """
    if sum(obtained_credit) >= 150:
        return True

    return False
```

```python
# student data in student_data_2026.csv

data_file = "student_data_2026.csv"

try:
    with open(data_file, "r") as f:
        data = f.readlines()
    
        clean_data = clean_student_data(data[1:])

except FileNotFoundError as e:
    print("Error: {e}")

else:
    for i in clean_data:
        credits = obtained_credits(i)

        # for testing, to display final results
        # print(f"Student {i[0]} total credits: {sum(credits)}/180")
        # if credits_achieved(credits):
        #     print("Eligible to graduate")
        # else:
        #     print("Insufficient credits")

    # maybe you want to output it to another file called results.txt
    with open("results.txt", "w") as f:
        f.write("StudentId,Credit_Obtained,Grad_Eligibility\n")
        for i in clean_data:
            credits = obtained_credits(i)
            if credits_achieved(credits):
                f.write(f"{i[0]},{sum(credits)},Eligible\n")
            else:
                f.write(f"{i[0]},{sum(credits)},Not Eligible\n")
        
```
