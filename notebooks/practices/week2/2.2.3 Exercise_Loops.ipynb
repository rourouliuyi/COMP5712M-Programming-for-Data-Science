### Exercise 3: Reversing a nested loop pattern

```python
n=5
for i in range(n):
    # Initialise printed output as an empty string
    output=''
    # The inner `for` loop will do just j=0 on the first outer loop,
    # then j=0,1 on the second outer loop, j=0,1,2 on the third outer loop, etc.
    for j in range(i+1):
        # Want to separate the numbers with a comma after the first
        if j==0:
            sep=''
        else:
            sep=', '
        # Add the separator and j+1 to the output string
        output=output+sep+f'{j+1}'
    # Print the output string. Note that this happens in the code block for
    # the outer `for` loop.
    print(output)
```
