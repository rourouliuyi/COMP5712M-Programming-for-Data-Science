### Slicing - similar to lists but with additional functionality

```python
grades2 = np.array([36, 60, 54, 47, 5, 98, 47, 30])
grades2[1]
```

```python
grades2[1:5]
```

```python
# slicing in NumPy allows for reassignment
grades2 = np.array([36, 60, 54, 47, 5, 98, 47, 30])

# can use logical operators to arrays
# here, if grade < 50, update to 0
grades2[grades2<50] = 0
grades2
```

```python
grades3 = np.array([36, 60, 54, 47, 5, 98, 47, 30, 50])

# 1. Convert all grades safely into a brand new array
# if grade >= 50, update to 15, else 0
credit = np.where(grades3 >= 50, 15, 0)

# 2. Update only the last element if it passed
if credit[-1] == 15:
    credit[-1] = 60

credit
```

```python
# np.nan (not a number), standard placeholder when there is no value

grades4 = np.array([36, 60, 54, 47, 5, 98, 47, np.nan, np.nan])
credit = np.where(grades3 >= 50, 15, 0)

# 2. Update only the last element if it passed
if grades3[-1] >= 50:
    credit[-1] = 60

# help(np.nan)

credit
```

```python
# can use np.sum
np.sum(credit)
```

```python
# if using sum(), it will fail if there is np.nan in the array
sum(grades4)
```

```python
# np.sum() will fail as well
np.sum(grades4)
```

```python
# there is a np.nansum() that can be used
np.nansum(grades4)
```

### 2. Revisiting student data example

We do not normally do this but this is just a demonstration of what can be done if using NumPy

```python
# Load just the numbers (skipping the first column with usecols=range(1,10) and the header row with skip_header=1)
# Empty cells automatically become 'nan' (Not a Number)

# use np.genfromtxt() if file contains missing values
# if no missing values, can use np.loadtxt()

def load_grades_file(filename):
    grades = np.genfromtxt(filename, delimiter=",", skip_header=1, usecols=range(1,10))
    return grades
```

```python
# testing
grades = load_grades_file("student_data_2026.csv")
print(grades)
```

```python
# Load just the Student IDs as text strings
def load_ids_file(filename):
    ids = np.genfromtxt(filename, delimiter=',', skip_header=1, usecols=0, dtype=str)
    return ids
```

```python
# testing
ids = load_ids_file("student_data_2026.csv")
print(ids)
```

```python
def calculate_credit(grades):
    credit = np.where(grades >= 50, 15, 0)
    mask = (credit[:, -1] == 15)
    credit[mask, -1] = 60
    return credit
```

```python
# testing
credits = calculate_credit(grades)
print(credits)
```

```python
# calculate total credits
def calculate_total_credit(credit):
    return np.sum(credit, axis=1)
```

```python
# testing
total_credits = calculate_total_credit(credits)
print(total_credits)
```

```python
# zip ids and total_credits together
combined = zip(ids, total_credits)
# combined = np.column_stack((ids, total_credits))  # not ideal as all numbers become str

for i in combined:
    if i[1] >= 120:
        print(f"{i[0]} with {i[1]}/180 is eligible to graduate")
    else:
        print(f"{i[0]} with {i[1]}/180 is not eligible to graduate")
```

### Anything else we can analyse?

#### 1. Descriptive & exploratory analysis

- module performance: calculate the mean, median, and variance for each module to find the hardest and easiest module
- identify patterns in missing grades(blank cells) to flag potential dropouts or incomplete records

```python
# average grade overall
# nanmean to deal with np.nan
np.nanmean(grades)
```

```python
# average grade per module (column), axis=0 for column
np.nanmean(grades, axis=0)
```

```python
# average grade per student, axis=1 for row
np.nanmean(grades, axis=1)
```

```python
# calculate median for each module
print("---Module Performance---")
print("Median per module")
print(np.nanmedian(grades, axis=0))
print()

# calculate variance for each module
print("Variance per module")
print(np.nanvar(grades, axis=0))
print()

# calculate standard deviation for each module
print("Standard deviations per module")
print(np.nanstd(grades, axis=0))
print()
```

```python
# Missing value analysis
print("---Missing Value Analysis---")
print("Percentage of missing value per column")
print(np.isnan(grades).mean(axis=0) * 100)
print()

print("Number of missing value per column")
print(np.isnan(grades).sum(axis=0))
print()
```
